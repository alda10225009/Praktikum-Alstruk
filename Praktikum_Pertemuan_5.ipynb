```python
# Praktikum 1 - Mengenali Tipe Data

nama = "Fajar"
umur = 19
berat = 54,5
terdaftar = True

# Menampilkan nilai dan tipe data
print("Nama:", nama)
print(type(nama))

print("Umur:", umur)
print(type(umur))

print("Berat:", berat)
print(type(berat))

print("Terdaftar:", terdaftar)
print(type(terdaftar))

# Mengubah nilai umur menjadi 20
umur = 20

print("\nNama:", nama)
print(type(nama))

print("Umur:", umur)
print(type(umur))

print("Berat:", berat)
print(type(berat))

print("Terdaftar:", terdaftar)
print(type(terdaftar))
```

```python
# Praktikum 2 - Operator Aritmatika

harga = 12500
jumlah = 4

# Menghitung total harga
total_harga = harga * jumlah

# Menghitung sisa pembagian
sisa = total_harga % 2000

# Menghitung harga setelah diskon
diskon = 5000
harga_diskon = total_harga - diskon

print("Harga per buku: Rp", harga)
print("Jumlah buku:", jumlah)
print("Total harga: Rp", total_harga)
print("Sisa pembagian total harga dengan 2000: Rp", sisa)
print("Diskon: Rp", diskon)
print("Harga setelah diskon: Rp", harga_diskon)
```

```python
# Praktikum 3 - Operator Perbandingan dan Logika

nilai = 82
kehadiran = 90

# Memeriksa masing-masing syarat
syarat_nilai = nilai >= 75
print("Memenuhi syarat nilai minimal 75:", syarat_nilai)
syarat_kehadiran = kehadiran >= 80
print("Memenuhi syarat kehadiran minimal 80%:", syarat_kehadiran)

# Memeriksa kedua syarat menggunakan operator and
print("Memenuhi kedua syarat:", syarat_nilai and syarat_kehadiran)

# Mengubah nilai menjadi 70
nilai = 70
syarat_nilai = nilai >= 75
print("\nNilai diubah menjadi:", nilai)
print("Memenuhi syarat nilai minimal 75:", syarat_nilai)
print("Memenuhi kedua syarat:", syarat_nilai and syarat_kehadiran)

# Mencoba menggunakan operator or
print("Memenuhi salah satu syarat:", syarat_nilai or syarat_kehadiran)
```

```python
# Praktikum 4 - Operator Penugasan

poin = 100
print("Poin awal:", poin)

poin += 25
print("Poin setelah ditambah 25:", poin)

poin -= 10
print("Poin setelah dikurangi 10:", poin)

poin *= 2
print("Poin setelah dikali 2:", poin)

poin /= 2
print("Poin setelah dibagi 2:", poin)
```

```python
# Praktikum 5 - Struktur Data Dasar

nilai = [78, 85, 90, 85]
mahasiswa = {"nama":"Nadia", "kelas":"IF-A"}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

# Menampilkan nilai pertama dari list
print("Nilai pertama dari list:", nilai[0])

# Menampilkan nilai 95 ke list
nilai.append(95)
print("List setelah ditambah angka 95:", nilai)

# Menampilkan nama mahasiswa
print("Nama mahasiswa:", mahasiswa["nama"])

# Menampilkan koordinat pertama
print("Koordinat pertama:", koordinat[0])

# Menampilkan angka unik secara berurutan
print("Angka unik:", sorted(angka))
```

```python
# Tantangan Terintegrasi - Program Rekap Nilai

nama = "Alda Aulia Hafid"
nilai_tugas = 86
nilai_ujian = 90

# menghitung rata-rata dua nilai
rata_rata = (nilai_tugas + nilai_ujian) / 2

# Memeriksa status kelulusan
lulus = rata_rata >= 75

# Menampilkan hasil
print("=== PENGUJIAN PERTAMA ===")
print("Nama mahasiswa:", nama)
print("Nilai Tugas:", nilai_tugas)
print("Nilai Ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status Lulus:", lulus)

# Uji program kedua
nama = "Alda Aulia Hafid"
nilai_tugas = 55
nilai_ujian = 60

rata_rata = (nilai_tugas + nilai_ujian) / 2
lulus = rata_rata >= 75

print("\n==== PENGUJIAN KEDUA ====")
print("Nama mahasiswa:", nama)
print("Nilai Tugas:", nilai_tugas)
print("Nilai Ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status Lulus:", lulus)
```
