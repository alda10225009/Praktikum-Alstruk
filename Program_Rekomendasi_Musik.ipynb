```python
# TUGAS PERTEMUAN 6
# Studi Kasus: Rekomendasi Musik Berdasarkan Genre & Mood

# Menerima input dari pengguna (dibuat huruf kecil agar tidak sensitif huruf kapital)
print("=== WELCOME TO SMART MUSIC RECOMMENDER ===")
genre = input("Pilih genre (hiphop / rnb): ").lower()
mood = input("Bagaimana mood Anda saat ini? (energetic / chill): ").lower()

print("\n--- REKOMENDASI UNTUK ANDA ---")

# PERCABANGAN TINGKAT 1 (LUAR): FILTER GENRE
if genre == "hiphop":
  # PERCABANGAN TINGKAT 2 (DALAM): FILTER MOOD UNTUK HIP-HOP
    if mood == "energetic":
        print("Kendrick Lamar - 'Humble' atau Cortis Ft. Juicy J - 'Motion'")
    elif mood == "chill":
        print("Kendrick Lamar Ft. SZA - 'Luther' atau The Weeknd Ft. Playboi Carti 'Timeless'")
    else:
        print("Maaf, pilihan mood yang anda masukkan belum tersedia pada sistem Hip-Hop kami")

# PERCABANGAN TINGKAT 1 (LUAR): FILTER GENRE
elif genre == "rnb":
  # PERCABANGAN TINGKAT 2 (DALAM): FILTER MOOD UNTUK R&B
    if mood == "energetic":
        print("Bruno Mars - 'That's What i Like' atau Tinashe Ft. ScHoolboyQ - '2 On'")
    elif mood == "chill":
        print("Chase Atlantic - 'Friends' atau Daniel Caesar - 'Always'")
    else:
        print("Maaf, pilihan mood yang anda masukkan belum tersedia pada sistem R&B kami")

else:
    print("Maaf, genre yang Anda masukkan belum tersedia di sistem kami.")
```
