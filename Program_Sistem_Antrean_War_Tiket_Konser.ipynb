```python
# =====================================================================
# TUGAS PERTEMUAN 8 ALGORITMA & STRUKTUR DATA
# NAMA: Alda Aulia Hafid
# NIM: 10225009
# STUDI KASUS: SISTEM ANTREAN WAR TIKET KONSER K-POP
# =====================================================================

# 1. TUPLE: Menyimpan data tetap (immutable) untuk informasi detail kategori tiket dan benefitnya
# Format: (Nama Section, Benefit / Fasilitas, Harga Satuan)
tiket_vip_sc = ("VIP Soundcheck", "Akses Soundcheck + Exclusive Photocard Set + Designated Seat", 3500000)
tiket_cat1 = ("CAT 1", "Numbered Seating (Pandangan Utama Dekat Main Stage)", 2400000)
tiket_cat2 = ("CAT 2", "Festival / Standing Area (Berdiri Bebas di Belakang CAT 1)", 1600000)
tiket_tribune = ("Tribune", "Numbered Seating (Tribun Atas / Panorama View)", 950000)

# 2. ARRAY / LIST: Menyimpan kumpulan data kategori tiket konser dalam satu struktur sekuensial
daftar_kategori_tiket = [tiket_vip_sc, tiket_cat1, tiket_cat2, tiket_tribune]

# Menggunakan operasi fungsi Array: append() untuk menambah kategori tiket baru (misal: Soundcheck Pass tambahan)
tiket_soundcheck_pass = ("Soundcheck Pass", "Special Entry + Official Lanyard + Poster", 1200000)
daftar_kategori_tiket.append(tiket_soundcheck_pass)


# 3. LINKED LIST: Struktur data linear dinamis untuk mengelola antrean "War Tiket" para fans
class NodeFans:
    def __init__(self, nama_fans, kategori_pilihan, jumlah_tiket):
        self.nama_fans = nama_fans                # Data: Nama Pembeli
        self.kategori_pilihan = kategori_pilihan  # Data: Kategori Tiket yang di-war
        self.jumlah_tiket = jumlah_tiket          # Data: Jumlah Tiket
        self.next = None                          # Pointer: Menunjuk ke antrean berikutnya (Singly Linked List)

class AntreanWarTiket:
    def __init__(self):
        self.head = None                          # Pointer awal antrean (Head)

    def tambah_antrean(self, nama_fans, kategori_pilihan, jumlah_tiket):
        """Menambahkan fans baru ke akhir antrean 'War Tiket' (Enqueue)"""
        node_baru = NodeFans(nama_fans, kategori_pilihan, jumlah_tiket)
        if not self.head:
            self.head = node_baru
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = node_baru
        print(f"[✨] Berhasil Masuk Antrean War: {nama_fans} mengincar tiket [{kategori_pilihan}]")

    def proses_pembelian(self):
        """Memproses dan menghapus antrean terdepan yang berhasil checkout (Dequeue - FIFO)"""
        if not self.head:
            print("\n[-] Antrean War Tiket Kosong! Semua tiket sudah terjual.")
            return

        pembeli_berhasil = self.head
        self.head = self.head.next  # Memindahkan head pointer ke node berikutnya

        # Mencari harga & benefit dari struktur Array & Tuple
        harga_satuan = 0
        benefit_info = ""
        for kat in daftar_kategori_tiket:
            if kat[0].lower() == pembeli_berhasil.kategori_pilihan.lower():
                benefit_info = kat[1]
                harga_satuan = kat[2]

        total_harga = harga_satuan * pembeli_berhasil.jumlah_tiket

        print(f"\n🎟️ [STATUS CHECKOUT WAR TIKET BERHASIL!] 🎟️")
        print(f"    Nama          : {pembeli_berhasil.nama_fans}")
        print(f"    Section Tiket : {pembeli_berhasil.kategori_pilihan}")
        print(f"    Benefit       : {benefit_info}")
        print(f"    Jumlah Tiket  : {pembeli_berhasil.jumlah_tiket} Pcs")
        print(f"    Total Tagihan : Rp {total_harga:,}")
        print(f"    Status        : Pembayaran Terkonfirmasi & E-Ticket Terbit! 💖\n")

    def tampilkan_antrean(self):
        """Menampilkan daftar antrean war tiket saat ini (Traversal Linked List)"""
        current = self.head
        if not current:
            print("\n✨ [Info Antrean]: Saat ini antrean war sudah bersih.\n")
            return

        print("\n" + "="*65)
        print("         DAFTAR TUNGGU ANTREAN 'WAR TIKET' K-POP KONSER        ")
        print("="*65)
        no = 1
        while current:
            print(f"{no}. Nama: {current.nama_fans:<18} | Section: {current.kategori_pilihan:<16} | Qty: {current.jumlah_tiket} Tiket")
            current = current.next
            no += 1
        print("="*65 + "\n")


# ==========================================
# PROGRAM UTAMA (MAIN PROGRAM)
# ==========================================
if __name__ == "__main__":
    print("="*65)
    print("      CONCERT TICKET SYSTEM: TREASURE WORLD TOUR in JKT       ")
    print("="*65 + "\n")

    # 1. Menampilkan Katalog Menu Tiket (Menggunakan ARRAY & TUPLE)
    print("--- 1. KATALOG SECTION & BENEFIT TIKET ---")
    for index, kategori in enumerate(daftar_kategori_tiket):
        print(f"[{index}] Section : {kategori[0]}")
        print(f"    Benefit : {kategori[1]}")
        print(f"    Harga   : Rp {kategori[2]:,}")
        print("-" * 60)

    # Menggunakan fungsi len() pada Array
    print(f"Total Pilihan Section Tiket Konser: {len(daftar_kategori_tiket)} Jenis\n")

    # 2. Simulasi Antrean War Tiket (Menggunakan LINKED LIST)
    print("--- 2. SIMULASI ANTREAN 'WAR TIKET' ---")
    sistem_antrean = AntreanWarTiket()

    # Menambahkan data antrean fans
    sistem_antrean.tambah_antrean("Alda", "VIP Soundcheck", 2)
    sistem_antrean.tambah_antrean("Sarah", "CAT 1", 1)
    sistem_antrean.tambah_antrean("Aurel", "CAT 2", 3)

    # Menampilkan antrean saat ini
    sistem_antrean.tampilkan_antrean()

    # 3. Proses Checkout War Tiket (Menghapus node terdepan)
    print("--- 3. PROSES CHECKOUT & PEMBAYARAN WAR TIKET ---")
    sistem_antrean.proses_pembelian()

    # Menampilkan sisa antrean setelah satu pembeli berhasil checkout
    sistem_antrean.tampilkan_antrean()
```
